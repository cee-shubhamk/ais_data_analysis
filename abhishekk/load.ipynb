```python
import polars as pl
from pathlib import Path

# Load parquet
file_path = Path("C:/Users/developer_02/Desktop/data_AIS/data_inst-dd_mp_etl02.parquet")
df = pl.read_parquet(file_path)

# Destination frequency table with string length
destination_frequency = (
    df.group_by("destination")
      .agg(
          pl.len().alias("occurrence_count")
      )
      .drop_nulls("destination")
      .with_columns(
          pl.col("destination")
            .str.len_chars()
            .alias("character_length")
      )
      .sort("occurrence_count", descending=True)
)

# Save CSV
output_file = Path("C:/Users/developer_02/Desktop/data_AIS/destination_frequency_table.csv")
destination_frequency.write_csv(output_file)

print("=" * 60)
print("DESTINATION FREQUENCY TABLE CREATED")
print("=" * 60)
print(f"Total Unique Destinations : {destination_frequency.height:,}")
print(f"CSV Saved At              : {output_file}")
```
